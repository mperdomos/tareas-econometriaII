# Tarea 2: Ecuaciones de Yule-Walker para un AR(3) usando la Regla de Cramer

## Planteamiento del Ejercicio
Se analiza el proceso autorregresivo de orden 3, $AR(3)$:

$$X_t = \frac{1}{2}X_{t-1} + \frac{1}{10}X_{t-2} - \frac{1}{5}X_{t-3} + Z_t \quad \text{donde } Z_t \sim \text{RB}(0, \sigma_Z^2 = 1)$$

---

## 1. Sistema de Ecuaciones de Yule-Walker
Multiplicando la ecuación por $X_{t-h}$ y tomando esperanza matemática $E[\cdot]$:

* Para $h = 0$:
$$\gamma_0 = \frac{1}{2}\gamma_1 + \frac{1}{10}\gamma_2 - \frac{1}{5}\gamma_3 + 1 \quad \text{(Ecuación 1)}$$

* Para $h = 1$:
$$\gamma_1 = \frac{1}{2}\gamma_0 + \frac{1}{10}\gamma_1 - \frac{1}{5}\gamma_2 \implies -\frac{1}{2}\gamma_0 + \frac{9}{10}\gamma_1 - \frac{1}{5}\gamma_2 = 0 \quad \text{(Ecuación 2)}$$

* Para $h = 2$:
$$\gamma_2 = \frac{1}{2}\gamma_1 + \frac{1}{10}\gamma_0 - \frac{1}{5}\gamma_1 \implies -\frac{1}{10}\gamma_0 - \frac{3}{10}\gamma_1 + \gamma_2 = 0 \quad \text{(Ecuación 3)}$$

* Para $h = 3$:
$$\gamma_3 = \frac{1}{2}\gamma_2 + \frac{1}{10}\gamma_1 - \frac{1}{5}\gamma_0 \quad \text{(Ecuación 4)}$$

---

## 2. Reducción a un Sistema $3 \times 3$
Sustituyendo la Ecuación (4) en la Ecuación (1):

$$\gamma_0 = \frac{1}{2}\gamma_1 + \frac{1}{10}\gamma_2 - \frac{1}{5}\left(\frac{1}{2}\gamma_2 + \frac{1}{10}\gamma_1 - \frac{1}{5}\gamma_0\right) + 1$$

$$\gamma_0 = \frac{12}{25}\gamma_1 + \frac{1}{25}\gamma_0 + 1 \implies \frac{24}{25}\gamma_0 - \frac{12}{25}\gamma_1 = 1 \implies 24\gamma_0 - 12\gamma_1 = 25 \quad \text{(Ecuación 5)}$$

Obtenemos la forma matricial $A \mathbf{\gamma} = \mathbf{b}$:

$$\begin{pmatrix} 24 & -12 & 0 \\ -5 & 9 & 2 \\ -1 & -3 & 10 \end{pmatrix} \begin{pmatrix} \gamma_0 \\ \gamma_1 \\ \gamma_2 \end{pmatrix} = \begin{pmatrix} 25 \\ 0 \\ 0 \end{pmatrix}$$

---

## 3. Solución Exacta mediante Regla de Cramer
Calculamos los determinantes:
* $\det(A) = 24(90 + 6) - 12(-50 + 2) + 0 = 2304 - 576 = 1728$
* $\det(A_0) = 25(90 + 6) = 2400 \implies \gamma_0 = \frac{2400}{1728} = \frac{25}{18}$
* $\det(A_1) = -25(-50 + 2) = 1200 \implies \gamma_1 = \frac{1200}{1728} = \frac{25}{36}$
* $\det(A_2) = 25(15 - (-9)) = 600 \implies \gamma_2 = \frac{600}{1728} = \frac{25}{72}$

Sustituyendo en la Ecuación (4):
$$\gamma_3 = \frac{1}{2}\left(\frac{25}{72}\right) + \frac{1}{10}\left(\frac{25}{36}\right) - \frac{1}{5}\left(\frac{25}{18}\right) = -\frac{5}{144}$$


```python
import numpy as np

# 1. Matriz de coeficientes A y vector b
A = np.array([
    [24, -12, 0],
    [-5, 9, 2],
    [-1, -3, 10]
], dtype=float)

b = np.array([25, 0, 0], dtype=float)

# 2. Determinante principal det(A)
det_A = np.linalg.det(A)

# 3. Determinantes sustituyendo columnas (Regla de Cramer)
A0, A1, A2 = A.copy(), A.copy(), A.copy()
A0[:, 0] = b
A1[:, 1] = b
A2[:, 2] = b

det_A0 = np.linalg.det(A0)
det_A1 = np.linalg.det(A1)
det_A2 = np.linalg.det(A2)

# Valores exactos
gamma_0_teorico = det_A0 / det_A
gamma_1_teorico = det_A1 / det_A
gamma_2_teorico = det_A2 / det_A
gamma_3_teorico = 0.5 * gamma_2_teorico + 0.1 * gamma_1_teorico - 0.2 * gamma_0_teorico

print("=== RESULTADOS TEÓRICOS (REGLA DE CRAMER) ===")
print(f"det(A)  = {det_A:.0f}")
print(f"det(A0) = {det_A0:.0f}  ->  gamma_0 = {gamma_0_teorico:.6f}  (Fracción: 25/18)")
print(f"det(A1) = {det_A1:.0f}  ->  gamma_1 = {gamma_1_teorico:.6f}  (Fracción: 25/36)")
print(f"det(A2) = {det_A2:.0f}  ->  gamma_2 = {gamma_2_teorico:.6f}  (Fracción: 25/72)")
print(f"gamma_3 = {gamma_3_teorico:.6f}  (Fracción: -5/144)")
```

## 4. Verificación por Simulación Numérica en Python
Simulamos una serie $AR(3)$ de tamaño $N = 100,000$ para comprobar que las autocovarianzas muestrales convergen a los valores teóricos.


```python
# Simulación del proceso AR(3)
np.random.seed(2024)
N = 100000
z = np.random.normal(0, 1, N)
x = np.zeros(N)

phi1, phi2, phi3 = 0.5, 0.1, -0.2

for t in range(3, N):
    x[t] = phi1 * x[t-1] + phi2 * x[t-2] + phi3 * x[t-3] + z[t]

# Descartar periodo de calentamiento (burn-in)
x_sim = x[1000:]

# Autocovarianzas muestrales
def autocovariance(x, lag):
    x_mean = np.mean(x)
    return np.mean((x[:len(x)-lag] - x_mean) * (x[lag:] - x_mean))

g0_muestral = autocovariance(x_sim, 0)
g1_muestral = autocovariance(x_sim, 1)
g2_muestral = autocovariance(x_sim, 2)
g3_muestral = autocovariance(x_sim, 3)

print("=== COMPARACIÓN: TEÓRICO vs MUESTRAL ===")
print(f"gamma_0: Teórico = {gamma_0_teorico:.6f} | Muestral = {g0_muestral:.6f}")
print(f"gamma_1: Teórico = {gamma_1_teorico:.6f} | Muestral = {g1_muestral:.6f}")
print(f"gamma_2: Teórico = {gamma_2_teorico:.6f} | Muestral = {g2_muestral:.6f}")
print(f"gamma_3: Teórico = {gamma_3_teorico:.6f} | Muestral = {g3_muestral:.6f}")
```
