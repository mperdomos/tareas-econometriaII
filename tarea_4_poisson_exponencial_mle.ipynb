# Tarea 4: Estimación por Máxima Verosimilitud - Poisson y Exponencial

**Objetivo:** Definir las distribuciones de Poisson y Exponencial, estimar analíticamente su parámetro $\lambda$ por Máxima Verosimilitud e interpretar el resultado.

## 1. Distribución de Poisson

La distribución de Poisson modela el número de eventos en un intervalo fijo de tiempo o espacio.
* **PMF:** $P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}, \quad k = 0, 1, 2, \dots$
* **Log-Verosimilitud:** $\ln L(\lambda) = \sum_{i=1}^n x_i \ln\lambda - n\lambda - \sum_{i=1}^n \ln(x_i!)$
* **CPO:** $\frac{d \ln L}{d \lambda} = \frac{\sum x_i}{\lambda} - n = 0 \implies \hat{\lambda}_{Poisson} = \bar{X}$

---

## 2. Distribución Exponencial

La distribución Exponencial modela el tiempo transcurrido entre eventos de un proceso de Poisson.
* **PDF:** $f(x; \lambda) = \lambda e^{-\lambda x}, \quad x \ge 0$
* **Log-Verosimilitud:** $\ln L(\lambda) = n \ln \lambda - \lambda \sum_{i=1}^n x_i$
* **CPO:** $\frac{d \ln L}{d \lambda} = \frac{n}{\lambda} - \sum_{i=1}^n x_i = 0 \implies \hat{\lambda}_{Exp} = \frac{1}{\bar{X}}$

### Interpretación de $\hat{\lambda}$:
* En **Poisson**, $\hat{\lambda}$ representa la **tasa o número promedio de ocurrencias** por unidad de tiempo.
* En **Exponencial**, $\hat{\lambda}$ representa la **tasa de eventos por unidad de tiempo**. Su recíproco $1/\hat{\lambda} = \bar{X}$ representa el **tiempo promedio esperado de espera** entre eventos consecutivas.

```python
import numpy as np

np.random.seed(2024)
n = 1000

# 1. Simulación Poisson
lambda_poisson_real = 4.5
data_poisson = np.random.poisson(lam=lambda_poisson_real, size=n)
lambda_poisson_mle = np.mean(data_poisson)

# 2. Simulación Exponencial
lambda_exp_real = 0.8  # tasa (eventos por unidad de tiempo)
# En numpy scale = 1/lambda
data_exp = np.random.exponential(scale=1.0/lambda_exp_real, size=n)
lambda_exp_mle = 1.0 / np.mean(data_exp)

print("=== ESTIMACIONES DE MÁXIMA VEROSIMILITUD ===")
print(f"Poisson:     Lambda Verdadero = {lambda_poisson_real:.2f} | Estimado MLE (X_bar) = {lambda_poisson_mle:.4f}")
print(f"Exponencial: Lambda Verdadero = {lambda_exp_real:.2f} | Estimado MLE (1/X_bar) = {lambda_exp_mle:.4f}")
```
