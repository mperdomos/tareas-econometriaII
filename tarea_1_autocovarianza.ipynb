# Tarea 1: Autocovarianza de un Proceso Centrado

**Objetivo:** Demostrar formalmente cómo se pasa de $x_t = y_t - E[y_t]$ a la igualdad de autocovarianzas $\gamma_y(s) = r_x(s)$, explicando por qué es necesario centrar un proceso estocástico.

## 1. Demostración Teoríca Paso a Paso

Sea $\{y_t\}_{t \in \mathbb{Z}}$ un proceso estocástico estacionario en sentido débil con media constante $\mu = E[y_t]$.

Definimos el **proceso centrado** $x_t$ como:
$$x_t = y_t - E[y_t] = y_t - \mu$$

Notemos que el valor esperado de $x_t$ es cero:
$$E[x_t] = E[y_t - \mu] = E[y_t] - \mu = 0$$

### Paso a Paso para la Autocovarianza $r_x(s)$:
La autocovarianza del proceso centrado $x_t$ entre el periodo $t$ y $t-s$ se define como $E[x_t x_{t-s}]$ dado que $E[x_t]=0$:

$$r_x(s) = E[x_t x_{t-s}] = E[(y_t - E[y_t])(y_{t-s} - E[y_{t-s}])]$$

Sustituyendo $E[y_t] = E[y_{t-s}] = \mu$ e insertando en la multiplicación:
$$r_x(s) = E[(y_t - \mu)(y_{t-s} - \mu)]$$
$$r_x(s) = E[y_t y_{t-s} - \mu y_t - \mu y_{t-s} + \mu^2]$$

Aplicando la linealidad del operador esperanza $E[\cdot]$:
$$r_x(s) = E[y_t y_{t-s}] - \mu E[y_t] - \mu E[y_{t-s}] + \mu^2$$
$$r_x(s) = E[y_t y_{t-s}] - \mu(\mu) - \mu(\mu) + \mu^2$$
$$r_x(s) = E[y_t y_{t-s}] - \mu^2$$

Por definición, la **autocovarianza** del proceso original $y_t$ es:
$$\gamma_y(s) = \text{Cov}(y_t, y_{t-s}) = E[y_t y_{t-s}] - E[y_t]E[y_{t-s}] = E[y_t y_{t-s}] - \mu^2$$

Por lo tanto, concluimos que:
$$\gamma_y(s) = r_x(s)$$

**¿Por qué se centra un proceso?**
Centrar la serie restando su media constante $\mu$ simplifica el álgebra al eliminar términos constantes en los momentos centrados, facilitando el análisis de la estructura de dependencia temporal sin alterar la autocovarianza del proceso.

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulación numérica para verificar la teoría
np.random.seed(2024)
T = 1000
mu = 5.0
rho = 0.7
sigma = 1.0

# Generar proceso AR(1) con media mu
epsilon = np.random.normal(0, sigma, T)
y = np.zeros(T)
y[0] = mu
for t in range(1, T):
    y[t] = mu*(1-rho) + rho*y[t-1] + epsilon[t]

# Proceso centrado
x = y - np.mean(y)

# Calcular autocovarianzas muestrales para lags s = 0, 1, 2, 3, 4
def autocovariance(series, max_lag=5):
    n = len(series)
    mean = np.mean(series)
    res = []
    for s in range(max_lag + 1):
        cov = np.mean((series[:n-s] - mean) * (series[s:] - mean))
        res.append(cov)
    return np.array(res)

gamma_y = autocovariance(y, max_lag=5)
r_x = np.array([np.mean(x[:T-s] * x[s:]) for s in range(6)])

print("Lag (s) | Autocovarianza gamma_y(s) | Autocovarianza r_x(s) | Diferencia")
print("-" * 65)
for s in range(6):
    diff = abs(gamma_y[s] - r_x[s])
    print(f"   {s}    |        {gamma_y[s]:.6f}          |       {r_x[s]:.6f}      |  {diff:.2e}")
```
