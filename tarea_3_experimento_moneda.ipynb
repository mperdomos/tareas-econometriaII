# Tarea 3: Experimento Lanzamiento de Moneda y Máxima Verosimilitud

**Objetivo:** Definir el experimento de lanzar una moneda con probabilidades dadas, caracterizar la variable aleatoria asociada, su función de masa de probabilidad (PMF), función de distribución acumulada (CDF) y encontrar la estimación por Máxima Verosimilitud (MLE) para el pronóstico.

## 1. Definición del Experimento y Variable Aleatoria

* **Espacio Muestral:** $\Omega = \{\text{sello}, \text{cara}\}$.
* **Probabilidades:** $P(\text{sello}) = 0.65$, $P(\text{cara}) = 0.35$.
* **Variable Aleatoria $X$:** Definimos la variable dicotómica de Bernoulli:
  $$X(\omega) = \begin{cases} 1 & \text{si sale sello} \\ 0 & \text{si sale cara} \end{cases}$$

### Función de Masa de Probabilidad (PMF)
Para $\theta = P(X=1) = 0.65$:
$$f(x; \theta) = P(X=x) = \theta^x (1-\theta)^{1-x}, \quad x \in \{0, 1\}$$

### Función de Distribución Acumulada (CDF)
$$F(x) = P(X \le x) = \begin{cases} 0 & x < 0 \\ 1 - \theta = 0.35 & 0 \le x < 1 \\ 1 & x \ge 1 \end{cases}$$

### Estimador de Máxima Verosimilitud (MLE) para el Pronóstico
Dada una muestra de $N$ lanzamientos con $k = \sum_{i=1}^N x_i$ sellos:
$$L(\theta) = \prod_{i=1}^N \theta^{x_i} (1-\theta)^{1-x_i} = \theta^k (1-\theta)^{N-k}$$
$$\ln L(\theta) = k \ln(\theta) + (N-k) \ln(1-\theta)$$

**Condición de Primer Orden:**
$$\frac{d \ln L}{d \theta} = \frac{k}{\theta} - \frac{N-k}{1-\theta} = 0 \implies \hat{\theta}_{MLE} = \frac{k}{N} = \bar{X}$$

El valor $\hat{\theta}_{MLE}$ que maximiza la probabilidad del pronóstico es la frecuencia relativa observada de sellos.

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulación computacional de la moneda
np.random.seed(2024)
theta_real = 0.65
N = 500

# Generar lanzamientos
lanzamientos = np.random.choice([1, 0], size=N, p=[theta_real, 1 - theta_real])
k_sellos = np.sum(lanzamientos)
theta_mle = k_sellos / N

print(f"Lanzamientos totales (N): {N}")
print(f"Sellos obtenidos (k): {k_sellos}")
print(f"Estimador MLE theta_hat = k/N: {theta_mle:.4f} (Verdadero: {theta_real})")

# Evaluamos la log-verosimilitud para un grid de valores de theta
thetas = np.linspace(0.01, 0.99, 100)
log_lik = k_sellos * np.log(thetas) + (N - k_sellos) * np.log(1 - thetas)

print(f"El valor theta que maximiza la log-verosimilitud es: {thetas[np.argmax(log_lik)]:.2f}")
```
